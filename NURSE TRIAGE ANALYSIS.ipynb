```python
# 1. Convert to datetime if not already done
triage['Triage Record Date'] = pd.to_datetime(triage['Triage Record Date'])

# 2. Extract date and day of the week
triage['Triage Date'] = triage['Triage Record Date'].dt.date
triage['Triage Day'] = triage['Triage Record Date'].dt.day_name()

# 3. Group by nurse and date
unique_patients = triage.groupby(['Login ID', 'Triage Date'])['Patient No'].nunique().reset_index()
unique_patients.rename(columns={'Patient No': 'unique_patients'}, inplace=True)

total_triages = triage.groupby(['Login ID', 'Triage Date']).size().reset_index(name='total_triages')

# 4. Merge into summary
summary = pd.merge(unique_patients, total_triages, on=['Login ID', 'Triage Date'])

# 5. Add day of the week (join back from df)
summary = pd.merge(summary, triage[['Login ID', 'Triage Date', 'Triage Day']].drop_duplicates(),
                   on=['Login ID', 'Triage Date'], how='left')

# 6. Reorder columns if you wish
summary = summary[['Login ID', 'Triage Date', 'Triage Day', 'unique_patients', 'total_triages']]

# Show results
summary.head()
```
